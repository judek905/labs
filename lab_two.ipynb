## Measures of Central Tendency
They are the only three example:
- Mean
- Median
- Mode    

```python
import pandas as pd
student_data = {
    "Name":["JOHN", "JANE", "JAMES", "JACK", "JILL", "ANDREW"],
    "Science":[90, 85, 95, 80, 70, 75],
    "Math":[85, 90, 80, 75, 95, 100],
    "Gender":["Male", "Female", "Male", "Male", "Female", "Male"]
}
students = pd.DataFrame(student_data)
students
```

```python
students["Math"].mode()[0]
```

## ASSIGNMENT
- Find the mean science mark and explain your result
- Find ythe median of the science mark and list the student attatched to that mark

```python
students ["Science"].mean()
```

```python
students["Science"].median()
```

## MEASURES OF DISPERSION
- Standard deviation
- Varience
- Range

```python
## Standard Deviation
students["Math"].std()
```

```python
## Variance
students["Math"].var()
```

```python
## Range
students["Math"].max() - students["Math"].min()
```

## MEASURES OF POSITION
- percentiles
- Quatiles
- Deciles etc

```python
## Quartiles
### lower quantiles
lower_quantile = students["Math"].quantile(0.25)
lower_quantile
```

```python
### upper quantiles
upper_quantile = students["Math"].quantile(0.75)
upper_quantile
```

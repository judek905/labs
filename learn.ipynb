## 3. Built-in types and type conversion
Core types include `int`, `float`, `complex`, `str`, `bool`, and `NoneType`. Collections are introduced below.

````python
whole = 12
decimal = 2.5
complex_value = 1 + 2j
text = 'hello'
truth = True
nothing = None
print(type(whole), type(decimal), type(complex_value), type(text), type(truth), type(nothing))
print(int('42'), float('3.5'), str(whole), bool(1))
print(isinstance(whole, int))
````

## 4. Operators and comparisons
Python supports arithmetic, comparison, logical, membership, and identity operators. Use `==` for equality and `is None` when checking for `None`.

````python
a, b = 17, 5
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** 2)
print(a > b, a == b, a != b)
print(True and False, True or False, not True)
print('py' in 'python', 3 in [1, 2, 3])
maybe_value = None
print(maybe_value is None)
````

## 5. Strings and formatting
Strings are immutable Unicode sequences. Indexing starts at zero; slices use an exclusive end index. F-strings are a convenient formatting method.

````python
name = 'Ada Lovelace'
print(name[0], name[-1], name[0:3], len(name))
print(name.lower(), name.upper(), name.replace('Ada', 'Countess'))
parts = 'red,green,blue'.split(',')
print(parts, ' | '.join(parts))
print(f'{name} wrote about the Analytical Engine.')
print(f'{3.14159:.2f}')
````

## 6. Collections: lists, tuples, ranges, sets, and dictionaries
Choose a collection based on its behavior: mutable sequence (list), immutable sequence (tuple), unique items (set), or key-value mapping (dict).

````python
numbers = [3, 1, 4]
numbers.append(2)
numbers.sort()
print(numbers, numbers[0], numbers[-1], numbers[1:3])
coordinates = (4, 9)
x, y = coordinates
print(x, y, list(range(2, 8, 2)))
unique = {1, 2, 2, 3}
unique.add(4)
print(sorted(unique), unique | {4, 5})
person = {'name': 'Grace', 'role': 'programmer'}
person['city'] = 'New York'
print(person['name'], person.get('age', 'unknown'))
print(list(person.keys()), list(person.values()))
````

## 7. Conditions and pattern matching
Indentation defines blocks. `match`/`case` (Python 3.10+) can select behavior based on a value or structure.

````python
temperature = 18
if temperature >= 25:
    description = 'warm'
elif temperature >= 15:
    description = 'mild'
else:
    description = 'cool'
print(description)

command = 'start'
match command:
    case 'start':
        print('Starting')
    case 'stop':
        print('Stopping')
    case _:
        print('Unknown command')
````

## 8. Loops, iteration, and comprehensions
Use `for` to iterate over items and `while` when a condition controls repetition. `break` exits a loop; `continue` skips to the next iteration.

````python
for index, number in enumerate(numbers, start=1):
    print(index, number)

for key, value in person.items():
    print(key, value)

counter = 3
while counter > 0:
    print(counter)
    counter -= 1

squares = [n * n for n in range(10) if n % 2 == 0]
square_map = {n: n * n for n in range(4)}
print(squares, square_map)
````

## 9. Functions, parameters, and scope
Functions make reusable units of logic. Prefer returning values and keeping functions focused. Avoid mutable default parameters.

````python
def greet(person_name='friend'):
    '''Return a friendly greeting.'''
    return f'Hello, {person_name}!'

def describe(name, *, city='unknown'):
    return f'{name} lives in {city}.'

def total(*values):
    return sum(values)

def build_profile(**details):
    return details

double = lambda value: value * 2
print(greet(), greet('Lin'))
print(describe('Sam', city='Lima'), total(1, 2, 3), double(5))
print(build_profile(name='Ari', active=True))
````

## 10. Type hints and documentation
Type hints document intended types and help tools find mistakes. Python does not enforce them automatically at runtime.

````python
def average(values: list[float]) -> float:
    '''Return the arithmetic mean of a non-empty list.'''
    if not values:
        raise ValueError('values must not be empty')
    return sum(values) / len(values)

print(average([2.0, 4.0, 6.0]))
print(average.__doc__)
````

## 12. Exceptions and validation
Handle errors you can meaningfully recover from. Raise clear exceptions when a function receives invalid input.

````python
def parse_positive_integer(value):
    try:
        number = int(value)
    except (TypeError, ValueError) as error:
        raise ValueError('provide a whole number') from error
    if number <= 0:
        raise ValueError('number must be positive')
    return number

print(parse_positive_integer('12'))
try:
    parse_positive_integer('0')
except ValueError as error:
    print('Input error:', error)
````

## 14. Classes, objects, and inheritance
Classes describe objects that combine data and behavior. Use composition or inheritance where they clarify the design.

````python
class Animal:
    def __init__(self, name: str):
        self.name = name

    def speak(self) -> str:
        return 'sound'

class Cat(Animal):
    def speak(self) -> str:
        return 'meow'

pet = Cat('Milo')
print(pet.name, pet.speak(), isinstance(pet, Animal))
````

## 15. Dataclasses, enums, and properties
Dataclasses reduce boilerplate for data-focused classes. Properties provide attribute-style access to computed or validated values.

````python
from dataclasses import dataclass
from enum import Enum

@dataclass
class Point:
    x: float
    y: float

class Status(Enum):
    TODO = 'todo'
    DONE = 'done'

point = Point(2, 5)
print(point, Status.TODO.value)
````

## 16. Iterators and generators
An iterator produces values one at a time. Generators use `yield` and can efficiently represent large or unbounded sequences.

````python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1

print(list(count_up_to(5)))
print(next(iter(['first', 'second'])))
````

## 17. Decorators and context managers
Decorators wrap functions to add behavior. Context managers manage setup and cleanup, commonly through the `with` statement.

````python
from contextlib import contextmanager

def announce(function):
    def wrapper(*args, **kwargs):
        print('Calling', function.__name__)
        return function(*args, **kwargs)
    return wrapper

@announce
def add(a, b):
    return a + b

@contextmanager
def simple_resource():
    print('Acquire')
    try:
        yield 'resource'
    finally:
        print('Release')

print(add(2, 3))
with simple_resource() as resource:
    print('Using', resource)
````

## 18. Functional tools and sorting
Built-ins such as `sorted`, `zip`, and `map` help transform and combine data. Comprehensions are often the clearest option for simple transformations.

````python
students = [('Ari', 88), ('Bo', 95), ('Cy', 82)]
print(sorted(students, key=lambda student: student[1], reverse=True))
names = ['Ari', 'Bo']
scores = [88, 95]
print(list(zip(names, scores)))
print(list(map(str.upper, names)))
print(list(filter(lambda score: score >= 90, scores)))
````

## 19. Regular expressions and dates
For simple text operations prefer string methods. Regular expressions are useful for pattern matching; they are not a full parser for complex formats.

````python
import re
from datetime import datetime, timedelta

sentence = 'Contact learner@example.com for details.'
emails = re.findall(r'[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}', sentence)
print(emails)
today = date.today()
print(today, today + timedelta(days=7))
parsed = datetime.strptime('2026-01-15', '%Y-%m-%d')
print(parsed.strftime('%B %d, %Y'))
````

## 22. Concurrency concepts
`async`/`await` supports cooperative asynchronous I/O. Threads and processes are alternatives with different tradeoffs. This simple example uses a coroutine and does not perform network access.

````python
import asyncio

async def delayed_message():
    await asyncio.sleep(0.01)
    return 'Finished asynchronously'

print(await delayed_message())
````
